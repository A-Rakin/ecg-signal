```python
import scipy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from scipy.datasets import electrocardiogram

# Load ECG data
ecg = electrocardiogram()

# Display basic information
print("ECG shape:", ecg.shape)
print("Sampling frequency: 360 Hz")
print("Duration:", len(ecg) / 360, "seconds")
print("First 10 samples:", ecg[:10])
```

```python
fs = 360
```

```python
num_samples = len(ecg)

duration_seconds = num_samples / fs

print("Number of samples:", num_samples)
print("Sampling frequency:", fs, "Hz")
print("Duration:", duration_seconds, "seconds")
print("Duration in minutes:", duration_seconds / 60)
```

```python
time = np.arange(len(ecg)) / fs
```

```python
print(time[:10])
print("Time of sample 360:", time[360], "seconds")
```

```python
plt.figure(figsize=(15, 4))

plt.plot(time, ecg)

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude (mV)")
plt.title("Raw ECG Signal")

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
start_time = 5
end_time = 10

start_index = int(start_time * fs)
end_index = int(end_time * fs)

segment = ecg[start_index:end_index]
segment_time = time[start_index:end_index]

plt.figure(figsize=(15, 4))

plt.plot(segment_time, segment)

plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude (mV)")
plt.title("Raw ECG Signal: 5 to 10 Seconds")

plt.grid(True)
plt.tight_layout()
plt.show()
```

**Numerical Statistics**

```python
print("Minimum amplitude:", np.min(ecg), "mV")
print("Maximum amplitude:", np.max(ecg), "mV")
print("Mean amplitude:", np.mean(ecg), "mV")
print("Standard deviation:", np.std(ecg), "mV")
```
